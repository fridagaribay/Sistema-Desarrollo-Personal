```python
import sqlite3
import ipywidgets as widgets
import pandas as pd
from IPython.display import display, clear_output
from datetime import datetime
```

```python
conn = sqlite3.connect('desarrollo_personal.db')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS finanzas (
    id_finanza INTEGER PRIMARY KEY AUTOINCREMENT,
    fecha TEXT,
    tipo TEXT,
    monto REAL,
    descripcion TEXT
)
""")

conn.commit()
conn.close()

print("✅ Base de datos y tabla 'finanzas' lista")
```
