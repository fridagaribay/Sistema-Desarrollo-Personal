```python
import sqlite3

# Conexión a la base de datos (se creará un archivo llamado desarrollo_personal.db)
conn = sqlite3.connect('desarrollo_personal.db')
cursor = conn.cursor()

# Crear la tabla de Finanzas basada en tu diseño previo
cursor.execute('''
CREATE TABLE IF NOT EXISTS finanzas (
    id_finanza INTEGER PRIMARY KEY AUTOINCREMENT,
    fecha DATE NOT NULL,
    tipo VARCHAR(20) NOT NULL,
    monto DECIMAL(10,2) NOT NULL,
    descripcion VARCHAR(200)
)
''')

conn.commit()
conn.close()

print("¡Base de datos y tabla 'Finanzas' creadas exitosamente!")
```
